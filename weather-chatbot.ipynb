### 2.1 Mapping inputs to responses using dictionaries.

We now use dictionaries to map user intents. Instead of writing a new *dictionary*, use the **ZIP** function on a list. 

```python
# Write a function that takes in precept categories and responses lists and returns a mapping between them.

# - Use ZIP function to combine two independent lists (precepts and responses) and create a new map between them.
# - Search on the ZIP function and how to use it in a for loop
# - Run the main loop and check that this function works

# ___ Write your code here _____


def zip_responses(precepts, responses):
    
    #Takes two lists: precepts and responses, and returns a dictionary
    # mapping each precept to its corresponding response using zip().
   
    mapping = {}
    for pre, res in zip(precepts, responses):
        mapping[pre] = res
    return mapping
```

```python
# Write a function to create precept-response pairs and return them as a Dictionary.
# Give the function a characteristic name (e.g., pattern_responses) 
# You can directly write these precept response pairs or try the ZIP function

# ___ Write your code here _____

def mappings():
    
    #Creates predefined precepts and responses, and returns a dictionary
    #of mapped responses using zip_responses function.
   
    precepts = ["hello", "how are you", "thanks", "what is your name"]
    responses = ["Hi there!", "I'm just a bot, but I'm good!", "You're welcome!", "I'm ChatGPT, your assistant."]
    return zip_responses(precepts, responses)


# Example usage in chatbot loop
if __name__ == "__main__":
    response_map = mappings()
    while True:
        user_input = input("You: ").lower()
        if user_input in response_map:
            print("Bot:", response_map[user_input])
        elif user_input == "bye":
            print("Bot: Goodbye!")
            break
        else:
            print("Bot: Sorry, I don't understand that.")
```

```python
# Write a function to create 5 RE patterns and return them as separate variables.
# Give the function a characteristic name (e.g., re_patterns) 
# ___ Write your code here _____
 

def re_patterns():
    greetings = r"\b(hello|hi|hey|howdy|greetings)\b"
    farewells = r"\b(bye|goodbye|see you|take care|later)\b"
    gratitude = r"\b(thanks|thank you|thx|much appreciated|cheers)\b"
    name_queries = r"\b(what's your name|who are you|your name please|tell me your name|identify yourself)\b"
    feelings = r"\b(how are you|how do you feel|how's it going|what's up|are you okay)\b"
    # New patterns
    name_pattern = r"\b(my name is|i am|i'm called|you can call me|my name’s)\s+(\w+)\b"
    color_pattern = (
        r"\b("
        r"my favourite color is|my favorite color is|"
        r"my favourite colour is|my favorite colour is|"
        r"i like|i love|"
        r"my favourite is|my favorite is|"
        r"i'm into|"
        r"\w+\s+is my favourite color|"
        r"\w+\s+is my favorite colour"
        r")\s+(\w+)\b"
    )
    return greetings, farewells, gratitude, name_queries, feelings, name_pattern, color_pattern

```

You will write functions for main loop, detecting patterns and generating responses to make chatbot code reusable.

```python
## Detect pattern - Write a function to detect RE patterns in user input and return response category.

## Create the detect_pattern function with one argument : user_input
## 1. Call the re_patterns function and store patterns in seperate variables
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, use the re.search to check if the pattern exists in user input 
##    2.2 If there is a match, return the response type (e.g., "greetings" or "bye" or "happy")
##    2.3 Note: Use <span style="color:red; font-size: 14px"> elif </span> to match multiple conditions
##    2.4 If none of the conditions, return "default" which will be the default response of the chatbot
##    2.5 <span style="color:red; font-size: 14px"> Note: use else for last conditon </span>


# Give the function a characteristic name (e.g., detect_pattern) 
# ___ Write your code here _____


def detect_pattern(user_input):
    global user_state
    greetings, farewells, gratitude, name_queries, feelings, name_pattern, color_pattern = re_patterns()

    if re.search(greetings, user_input):
        return "greetings"
    elif re.search(farewells, user_input):
        return "farewells"
    elif re.search(gratitude, user_input):
        return "gratitude"
    elif re.search(name_queries, user_input):
        return "name_queries"
    elif re.search(feelings, user_input):
        return "feelings"
    elif result := re.search(name_pattern, user_input):
        user_state["name"] = result.group(1).capitalize()
        return "name_response"
    elif result := re.search(color_pattern, user_input):
        user_state["color"] = result.group(2).lower()  # Second group contains the color
        return "color_response"
    else:
        return "default"
```

```python
def re_responses():
    return {
        "greetings": "Hello! How can I help you?",
        "farewells": "Goodbye! Have a great day!",
        "gratitude": "You're very welcome!",
        "name_queries": "I'm ChatGPT, your virtual assistant.",
        "feelings": "I'm just a bot, but I'm functioning well. Thanks!",
        "name_response": "Nice to meet you, {name}! How can I assist you today?",
        "color_response": "Great to know that your favourite color is {color}.",
        "default": "I'm sorry, I didn't understand that. Could you rephrase?"
    }
```

```python
## Chatbot response - Write a function to return a responses by indexing dictionary with response categories.

## Create the chatbot_response function with one argument : user_input
## 1. Call the detect_pattern function and store the response in variable (e.g., detected_type)
## 2. Call the re_responses function and store responses in a variable (e.g., re_response)
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, check if the detected_type is one of the keys of the dictionary
##    2.2 If there is a match, return the response from dictonary
##    2.3 Note: Use elif to match multiple conditions
##    2.4 If none of the conditions, use "default" response

# Give the function a characteristic name (e.g., detect_pattern) 
# ___ Write your code here _____

def chatbot_response(user_input):
    detected_type = detect_pattern(user_input.lower())
    response_dict = re_responses()

    # Fetch user details
    name = user_state.get("name")
    color = user_state.get("color")

    if detected_type == "name_response":
        response = response_dict[detected_type]
        return response.format(name=name)
    elif detected_type == "color_response":
        response = response_dict[detected_type]
        return response.format(color=color)
    elif detected_type in response_dict:
        return response_dict[detected_type]
    else:
        return response_dict["default"]
```

```python
## A reminder for what the main loop looked like.
def chat():
    print("Chatbot: Hello! How can I help you today?")
    while True:
        user_input = input("You: ")

        if user_input.lower() == "bye":
            print("Chatbot: Goodbye! Chat session ended.")
            break

        bot_response = chatbot_response(user_input)
        print(f"Chatbot: {bot_response}")

# Start chatting
chat()
```

#### Writing advanced regular expressions

Let's exploring advanced REs by creating two more patterns to match the name and favourite color of the user. 

As name, and color can be any value, pattern matching needs to be generic and requires using alphabets [a-zA-Z], words [\w] and special symbols [*,+,?].

```python
# Write a regular expression pattern for a name and favourite color.

# 1. Names typically start with a letter in upper case followed by a sequence of letters in lower case.
# Try writing a few use cases to ensure that the pattern works.

# 2. Colors typically start with a pattern - color is <color_name> or I like <color_name>.
# Try writing a few use cases to ensure that the pattern works.


def extract_info(text):
    """
    Extracts user's name and favorite color using advanced regular expressions.
    Returns a tuple: (name, color)
    """
    # Advanced patterns for name and favorite color
    name_pattern = r"\b(my name is|i am|i'm|name:|you can call me)\s+([A-Z][a-z]+)\b"
    color_pattern = r"\b(my favourite color is|my favorite color is|i like|i love|color:)\s+([a-zA-Z]+)\b"

    name_match = re.search(name_pattern, text, re.IGNORECASE)
    color_match = re.search(color_pattern, text, re.IGNORECASE)

    name = name_match.group(2) if name_match else None
    color = color_match.group(2) if color_match else None

    return name, color

# Predefined test strings (no user input)
test_statements = [
    "My name is Alice",
    "I am Bob",
    "You can call me Charlie",
    "Name: Daniel",
    "I'm Eve",
    "My favorite color is blue",
    "I like red",
    "My favourite color is yellow",
    "I love purple",
    "Color: green",
    "Nice to meet you",
    "What’s the weather today?"
]

# Processing all test statements
for text in test_statements:
    name, color = extract_info(text)
    
    print(f"Input: {text}")
    if name:
        print(f"Chatbot: Nice to meet you, {name}!")
    if color:
        print(f"Chatbot: {color.capitalize()} is a beautiful color!")
    if not name and not color:
        print("Chatbot: Hmm... I didn’t catch a name or color.")
    print("---")
```

### 3.3 Integrating file-based data into chatbot logic

```python
## A code template for storing responses from JSON file..

# def load_files():
#    pd.read_json('intents.json')
#    Type your code here.

# import json
import random

def load_files(filepath="intents.json"):
    """
    Loads intent patterns and responses from a JSON file and returns two dictionaries:
    - rgx2int: regex pattern → intent tag
    - int2res: intent tag → list of responses
    """
    with open(filepath, "r") as file:
        data = json.load(file)

    rgx2int = {}
    int2res = {}

    for intent in data["intents"]:
        tag = intent["tag"]
        patterns = intent["patterns"]
        responses = intent["responses"]

        regex_pattern = r"|".join([p.lower() for p in patterns])
        rgx2int[regex_pattern] = tag
        int2res[tag] = responses

    return rgx2int, int2res
```

```python
## A template for what the response generation code will look like..

def generate_response(user_input):
    
    rgx2int, int2res = load_files()
    
    # Checks for user input in rgx2int
    # Uses tag of rgx2int in int2res to get output
    
    # Print match and return true
    # or return False 
    
#     def generate_response(user_input):
#     rgx2int, int2res = load_files()

    user_input = user_input.lower()
    for pattern, intent in rgx2int.items():
        if re.search(pattern, user_input):
            response = random.choice(int2res[intent])
            print(f"Chatbot: {response}")
            return True

    return False

```

```python
## A reminder for what the main loop code will look like..

# Function to get user input and respond
def chatbot():

    # User welcome and termination instruction.
    print("Welcome! Press bye to exit")

    # Main loop:
    while True:
        
        user_input = input("You: ")
                
        if user_input == "bye":
            print("Chatbot: Thank you for your interaction")
            break
        
        response_found = generate_response(user_input)
                
        if response_found == False:
            print("I'm sorry, could you rephrase that again")
        
# Run the chatbot
chatbot()
```

Import required libraries, Imports Python standard libraries, NLTK tools, and timezone utilities.

Download necessary NLTK data and ensures required tokenizers, stopwords, and lemmatizer resources are available.

Creates stemmer, lemmatizer, and stopwords list. Initializes user name and location.


```python
import random
import string
import nltk
import pytz
from datetime import datetime
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords

# Download necessary NLTK datasets
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

# Load NLTK resources
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

# Global variable for user's name
user_name = "there"
```

Text Preprocessing Function

Cleans and tokenizes text: lowers case, removes punctuation/stopwords, stems, and lemmatizes.


```python
# Preprocess text with optional debug print
def preprocess_text(text, debug=False):
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    tokens = word_tokenize(text)
    filtered = [word for word in tokens if word not in stop_words]
    stemmed = [stemmer.stem(word) for word in filtered]
    lemmatized = [lemmatizer.lemmatize(word) for word in filtered]

    return ' '.join(lemmatized)
```

Response Generator

Matches preprocessed input to chatbot patterns and returns a personalized response.



```python
def generate_response(user_input, intents_json):
    user_processed = preprocess_text(user_input, debug=True)

    for intent in intents_json["intents"]:
        for pattern in intent["patterns"]:
            pattern_processed = preprocess_text(pattern, debug=False)
            if user_processed == pattern_processed:
                response = random.choice(intent["responses"])
                # Replace placeholders with actual name and location
                return response.replace("{name}", user_name).replace("{location}", user_location)

    return None
```
